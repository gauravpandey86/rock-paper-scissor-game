```python
import random 

data_list=["rock", "paper","scissor"]

user_input=input("enter your choise (rock, paper, scissor):- ")
computer_input = random.choice(data_list)

print(f"user input :- {user_input} : computer input :- {computer_input}")

if computer_input == user_input :
    print("match tie")
    
elif user_input == "rock":
    if computer_input == "paper":
        print("computer win the match")
    else:
        print("user win the match")

elif user_input == "paper":
    if computer_input == "rock":
        print("user win the match")
    else:
        print("computer win the match")

elif user_input == "scissor":
    if computer_input == "rock":
        print("computer win the match")
    else:
        print("user win the match")
```
